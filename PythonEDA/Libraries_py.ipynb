#Practicing python libraries

```python
#importing
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
#scatterplot

xscatter = np.random.rand(50) * 10
yscatter = np.random.rand(50) * 10
colors = np.random.rand(50)
plt.figure(figsize=(6,4), facecolor='lightgray',edgecolor='blue')
plt.scatter(xscatter,yscatter,c=colors,alpha=0.7)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
#barplot

categories = ['Cat A','Cat B','Cat C','Cat D']
values = [12,13,43,19]
plt.figure(figsize=(6,2))
plt.bar(categories,values,color=['skyblue', 'lightcoral', 'lightgreen', 'gold'])
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#piechart

print("\n--- Pie Chart ---")
labels_pie = ['Apples', 'Bananas', 'Cherries', 'Dates']
sizes_pie = [15, 30, 45, 10]
explode_pie = (0, 0.5, 0, 0)  # Explode the 2nd slice (Bananas)
plt.figure(figsize=(6, 6))
plt.pie(sizes_pie, explode=explode_pie, labels=labels_pie, autopct='%1.1f%%', shadow=True, startangle=90)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.title("Pie Chart Example")
plt.show()
```

```python
#lineplot

print("\n--- Multiple Line Plot ---")
x_line = np.arange(0, 10, 0.5)
y_line1 = np.sin(x_line)
y_line2 = np.cos(x_line)
plt.figure(figsize=(8, 5))
plt.plot(x_line, y_line1, label='Sin Wave', color='red', linestyle='-', marker='o', markersize=4)
plt.plot(x_line, y_line2, label='Cos Wave', color='green', linestyle='--', marker='x', markersize=4)
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Multiple Line Plot Example")
plt.legend()
plt.grid(True)
plt.show()
```

#Seaborn

```python
tips = sns.load_dataset("tips")
```

```python
plt.figure(figsize=(8, 6))
sns.relplot(x='total_bill', y='tip' , hue='smoker',data=tips, kind='scatter',style='smoker')
plt.title("Relational Plot: Total Bill vs Tip (by Smoker)")
```

```python
plt.figure(figsize=(8, 6))
sns.catplot(x='day', y='total_bill', hue='smoker', data=tips, kind='box')
plt.title("Categorical Plot: Total Bill by Day and Smoker (Box Plot)")
plt.show()
```

```python
#count plot

plt.figure(figsize=(6, 4))
sns.countplot(x='sex', data=tips)
plt.title("Count Plot: Number of Diners by Sex")
plt.show()
```
